# EDA: parallel corpus

Look at language pair distribution, sentence length, basic dedup signals.

```python
import json
from pathlib import Path
from collections import Counter
import matplotlib.pyplot as plt
```

```python
rows = [json.loads(l) for l in Path('../data/processed/en_hi.train.jsonl').read_text(encoding='utf-8').splitlines() if l.strip()]
print(len(rows), 'pairs')
```

```python
pairs = Counter((r['src_lang'], r['tgt_lang']) for r in rows)
for k, v in pairs.most_common():
    print(k, v)
```

```python
src_lens = [len(r['src_text'].split()) for r in rows]
tgt_lens = [len(r['tgt_text'].split()) for r in rows]
fig, ax = plt.subplots(1, 2, figsize=(10, 4))
ax[0].hist(src_lens, bins=40)
ax[0].set_title('src length')
ax[1].hist(tgt_lens, bins=40)
ax[1].set_title('tgt length')
plt.tight_layout()
```
